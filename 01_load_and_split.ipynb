# Cargar y dividir datos

```python
import os
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
```

# Parámetros

```python
RND_SEED = 123
PCT_TEST = 0.2
TARGET_MAP = {"F": 0, "NF": 1}
# Paths
path_raw = os.path.join("data", "raw")
path_interim = os.path.join("data", "interim")

# Input
file_input = "New%20Spanish%20Academic%20Dataset.csv"

# Output
file_prep = "prep.csv"
file_train = "train.csv"
file_test = "test.csv"
```

# Cargar datos

```python
path_data_raw = os.path.join(path_raw, file_input)

df_raw = pd.read_csv(path_data_raw)
df_raw.head()
```

```python
print(f"{df_raw.shape=}")  # (número de filas, número de columnas)
```

```python
print(f"Distribución de clases en FINAL_LABEL:\n{df_raw['FINAL_LABEL'].value_counts()}")
print(f"\nProporción de clases (%):\n{df_raw['FINAL_LABEL'].value_counts(normalize=True) * 100}")
```

# Transformar datos

```python
df_prep = df_raw.copy()
df_prep['y_is_nf'] = df_prep['FINAL_LABEL'].map(TARGET_MAP)
df_prep['x_text'] = df_prep['REQUIREMENT']
df_prep = df_prep[["x_text", "y_is_nf"]]
```

# Dividir datos

```python
# Usar test_size y random_state parametrizados con estrategia de muestreo estratificada
# cuando la entrada es un objeto único, la salida mantendrá la misma estructura
df_train, df_test = train_test_split(
    df_prep,
    test_size=PCT_TEST,
    random_state=RND_SEED,
    stratify=df_prep['y_is_nf']
)

print(f"{df_train.shape=}")
print(f"{df_test.shape=}")
print(f"\nProporción de clases en train:\n{df_train['y_is_nf'].value_counts(normalize=True)}")
print(f"\nProporción de clases en test:\n{df_test['y_is_nf'].value_counts(normalize=True)}")
```

# Escribir datos

```python
if not os.path.exists(path_interim):
    print("¡Por favor, crea la carpeta /data/interim/!")

df_prep.to_csv(
    os.path.join(path_interim, file_prep),
    index=False)

df_train.to_csv(
    os.path.join(path_interim, file_train),
    index=False)

df_test.to_csv(
    os.path.join(path_interim, file_test),
    index=False)
```
